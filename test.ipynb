```python
df['Total Spend'] = df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] + df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds']
```

```python
df['Total Purchases'] = df['NumCatalogPurchases'] + df['NumStorePurchases'] + df['NumWebPurchases']
print(df['Total Purchases'])
```

```python
df['ROI'] = ((df['Z_Revenue'] - df['Z_CostContact']) / df['Z_CostContact']) * 100
print(df['ROI'])
```

```python
df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
print(df["Total_Children"])    # Count the total number of children in each household
```

```python
df["Income_Per_Child"] = df["Income"] / (df["Total_Children"] + 1) # Calculate income per child, adding 1 to avoid division by zero
print(df["Income_Per_Child"])
```

```python
campaign_columns = ["AcceptedCmp1","AcceptedCmp2","AcceptedCmp3","AcceptedCmp4","AcceptedCmp5","Response"]  

df[campaign_columns].sum() # Sum of accepted campaigns per customer
```

```python
df[campaign_columns].sum().plot(kind="bar",figsize=(8,5))   

plt.title("Campaign Acceptance")  
plt.xlabel("Campaign")
plt.ylabel("Number of Customers")

plt.show()
```

```python
sns.countplot(data=df, x="Education")
plt.show() 
```

```python
sns.pairplot(data=df,vars=["Income", "MntWines", "MntMeatProducts"])

plt.show()
```

1. Income vs Income

Shows the distribution of customer income. Most customers are concentrated at lower income values, with some high-income outliers.

2. MntWines vs MntWines

Shows the distribution of wine spending. Most customers spend relatively less, while a smaller number spend much more.

3. MntMeatProducts vs MntMeatProducts

Shows the distribution of meat-product spending. Most values are concentrated at lower spending levels with a long right tail.

4. Income vs MntWines

Shows the relationship between income and wine spending. There appears to be a positive relationship, with higher-income customers generally spending more.

5. Income vs MntMeatProducts

Shows the relationship between income and meat-product spending. The points suggest a positive association, but there is considerable variation.

6. MntWines vs Income

Same relationship as #4, with the axes reversed. It shows how wine spending varies with customer income.

7. MntWines vs MntMeatProducts

Shows the relationship between wine spending and meat-product spending. Customers spending more on wine often also show higher meat spending.

8. MntMeatProducts vs Income

Same relationship as #5, with the axes reversed. It shows how meat-product spending varies with income.

9. MntMeatProducts vs MntWines

Same relationship as #7, with the axes reversed. It shows the association between meat-product and wine spending.
